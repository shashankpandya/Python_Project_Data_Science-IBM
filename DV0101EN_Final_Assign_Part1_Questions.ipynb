# Final Assignment: Part 1 - Create Visualizations using Matplotlib, Seaborn & Folium
## Automobile Sales Analysis During Economic Recessions
XYZ Automotives - Data Visualization Project

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
import folium
print("Libraries imported successfully")
```

```python
# Load the dataset
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(URL)
print("Dataset loaded successfully")
```

## Task 1.1: Line chart showing automobile sales fluctuation year to year.

```python
df_line = df.groupby("Year")["Automobile_Sales"].mean().reset_index()
plt.figure(figsize=(12,6))
df_line.plot(x="Year", y="Automobile_Sales", kind="line", ax=plt.gca(), marker="o")
plt.xticks(list(range(1980, 2024)), rotation=75)
plt.xlabel("Year")
plt.ylabel("Automobile Sales")
plt.title("Automobile Sales Fluctuation Over the Years")
plt.tight_layout()
plt.show()
```

## Task 1.2: Line chart - Advertising Expenditure vs Automobile Sales during Non-Recession.

```python
df_no_recession = df[df["Recession"] == 0]
df_adv = df_no_recession.groupby("Year")[["Automobile_Sales", "Advertising_Expenditure"]].mean().reset_index()
fig, ax1 = plt.subplots(figsize=(12,6))
ax1.plot(df_adv["Year"], df_adv["Automobile_Sales"], color="blue", label="Automobile Sales", marker="o")
ax1.set_xlabel("Year")
ax1.set_ylabel("Automobile Sales", color="blue")
ax2 = ax1.twinx()
ax2.plot(df_adv["Year"], df_adv["Advertising_Expenditure"], color="orange", label="Advertising Expenditure", marker="s")
ax2.set_ylabel("Advertising Expenditure", color="orange")
plt.title("Advertising Expenditure vs Automobile Sales During Non-Recession")
fig.tight_layout()
plt.show()
```

## Task 1.3: Bar chart - Sales trend per vehicle type: Recession vs Non-Recession.

```python
df_bar = df.groupby(["Recession", "Vehicle_Type"])["Automobile_Sales"].mean().reset_index()
plt.figure(figsize=(12,6))
sns.barplot(data=df_bar, x="Vehicle_Type", y="Automobile_Sales", hue="Recession")
plt.title("Average Automobile Sales by Vehicle Type (Recession vs Non-Recession)")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Automobile Sales")
plt.legend(title="Recession", labels=["Non-Recession", "Recession"])
plt.tight_layout()
plt.show()
```

## Task 1.4: Subplot - GDP variations during recession vs non-recession.

```python
df_gdp_rec = df[df["Recession"] == 1].groupby("Year")["GDP"].mean().reset_index()
df_gdp_norec = df[df["Recession"] == 0].groupby("Year")["GDP"].mean().reset_index()
fig, axes = plt.subplots(1, 2, figsize=(12,8))
axes[0].plot(df_gdp_rec["Year"], df_gdp_rec["GDP"], color="red", marker="o")
axes[0].set_title("GDP During Recession Period")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("GDP")
axes[1].plot(df_gdp_norec["Year"], df_gdp_norec["GDP"], color="green", marker="o")
axes[1].set_title("GDP During Non-Recession Period")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("GDP")
plt.tight_layout()
plt.show()
```

## Task 1.5: Bubble plot - Impact of seasonality on automobile sales.

```python
size = df["Seasonality_Weight"] * 1000
plt.figure(figsize=(12,6))
scatter = plt.scatter(df["Month"], df["Automobile_Sales"], s=size, alpha=0.5, c=df["Recession"], cmap="RdYlGn")
plt.xlabel("Month")
plt.ylabel("Automobile Sales")
plt.title("Impact of Seasonality on Automobile Sales (Bubble Size = Seasonality Weight)")
plt.colorbar(scatter, label="Recession Period")
plt.tight_layout()
plt.show()
```

## Task 1.6: Scatter plot - Correlation between average vehicle price and sales volume during recessions.

```python
df_rec = df[df["Recession"] == 1]
plt.figure(figsize=(12,6))
plt.scatter(df_rec["Price"], df_rec["Automobile_Sales"], alpha=0.7, color="purple")
plt.xlabel("Average Vehicle Price")
plt.ylabel("Automobile Sales Volume")
plt.title("Correlation Between Average Vehicle Price and Sales Volume During Recessions")
plt.tight_layout()
plt.show()
```

## Task 1.7: Pie chart - Advertising expenditure during recession vs non-recession.

```python
adv_by_recession = df.groupby("Recession")["Advertising_Expenditure"].sum()
plt.figure(figsize=(8,6))
plt.pie(adv_by_recession, labels=["Non-Recession", "Recession"], autopct="%1.1f%%", startangle=90, colors=["#4CAF50", "#F44336"])
plt.title("Advertising Expenditure: Recession vs Non-Recession Periods")
plt.tight_layout()
plt.show()
```

## Task 1.8: Pie chart - Total advertising expenditure per vehicle type during recession.

```python
df_rec_vtype = df[df["Recession"] == 1].groupby("Vehicle_Type")["Advertising_Expenditure"].sum()
plt.figure(figsize=(8,6))
plt.pie(df_rec_vtype, labels=df_rec_vtype.index, autopct="%1.1f%%", startangle=90)
plt.title("Total Advertising Expenditure per Vehicle Type During Recession")
plt.tight_layout()
plt.show()
```

## Task 1.9: Line chart - Effect of unemployment rate on vehicle type and sales during recession.

```python
df_unemp = df[df["Recession"] == 1].groupby(["unemployment_rate", "Vehicle_Type"])["Automobile_Sales"].mean().reset_index()
plt.figure(figsize=(12,6))
for vtype in df_unemp["Vehicle_Type"].unique():
    data = df_unemp[df_unemp["Vehicle_Type"] == vtype]
    plt.plot(data["unemployment_rate"], data["Automobile_Sales"], label=vtype, marker="o")
plt.xlabel("Unemployment Rate")
plt.ylabel("Automobile Sales")
plt.title("Effect of Unemployment Rate on Vehicle Type Sales During Recession")
plt.legend(title="Vehicle Type")
plt.tight_layout()
plt.show()
```
